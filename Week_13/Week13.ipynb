```python
# NAME: journal.py

# IMPORT STATEMENTS

# NONE

# FUNCTION DEFINITIONS

# Function 1: write_entry(filename, entry)
def write_entry(filename, entry):

  # opens the file in append mode
  with open(filename, 'a') as file:

    # Writes the entry and ends with a newline character
    file.write(entry + '\n')

# Function 2: read_journal(filemane)
def read_journal(filename):

  try:
    # opens the file in read mode
    with open(filename, 'r') as file:
        contents = file.read()
        print("\n--- Your Entries ---\n")
        print(contents.strip())
        print("------------------------\n")
  except FileNotFoundError:
    print("No entries found.")

# Function 3: main()
def main():

  # Make the text file and print an introduction
  journal_file = 'my_journal.txt'
  print('Welcome to your personal journal!')

  # Ask the user 3 options: (W)rite, (R)ead, or 'quit'. while loops should be used.
  while True:
    choice = input('Do you want to (W)rite, (R)ead, or (Q)uit? ').strip().lower()

    if choice == 'quit':
      print('Goodbye!')
      break
    elif choice == 'w':
      thought = input('Enter your thought: ')
      write_entry(journal_file, thought)
      print("Entry saved!\n")
    elif choice == 'r':
      read_journal(journal_file)
    else:
      print("Invaild option. Please choose 'W' 'R' or 'quit'.\n")


# PROGRAM LOGIC

# Run the program logic
main()
```
