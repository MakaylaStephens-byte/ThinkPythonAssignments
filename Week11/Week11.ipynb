```python
# NAME: gradebook.py

# IMPORT STATEMENTS #
#(NONE)

# FUNCTION DEFINITIONS #

# 1.) Function: add_students. Parameter: gradebook.

def add_students(gradebook):

  # As long as the user is able to enter students's names.
  # The function stops, when 'done' is entered.


  while True:
    name = input("Enter student's name (or 'done')").strip()
    if name.lower() == 'done':
      break

  # As long as that the user can enter each student's (numeric) grades in.
  # grade range is 0-100
  # grades should also correspond with the given student.

    grade = float(input(f"Enter grade for {name} (0-100): "))
    if 0 <= grade <= 100:
        gradebook[name] = grade
    else:
      print("invalid grade. Please enter a numerical value. ")


# 2.) Function: calculate_average. Parameter: gradebook.
def calculate_average(gradebook):

  # Calculate and returns the average grade from the gradebook.
  if not gradebook:
    return 0.0
  total_grade = sum(gradebook.values())
  average_grade = total_grade / len(gradebook)
  return average_grade




# PROGRAM LOGIC #

# Add an empty dictionary (gradebook)

gradebook = {}

# Add names to the gradebook

add_students(gradebook)

# Print final report and class average
print("--- Class Report: ---")
for name, grade in gradebook.items():
  print(f"{name}: {grade}")

# Calculate and display class average under the report
if gradebook:
  average = calculate_average(gradebook)
  print(f"Class average: {average:.2f}")
```
