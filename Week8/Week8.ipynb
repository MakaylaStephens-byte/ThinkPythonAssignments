```python
### IMPORT STATEMENTS ###

### FUNCTION DEFINITIONS ###

# Create function, is_palindrom. Parameter: Word. Returns: True or False.
def is_palindrome(word):
    # Convert the word to lowercase (for case insensitivity)
    word = word.lower()
    # Compare the reversed version of the word.
    return word == word[::-1]

### PROGRAM LOGIC ###

# Ask user for a word.

# add value to word (for manually testing).
word = "Fork"
# Call the function.
is_palindrome(word)

# Print a user-friendly message for the result.
```
