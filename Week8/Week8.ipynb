```python
### IMPORT STATEMENTS ###

### FUNCTION DEFINITIONS ###

# Create function, is_palindrome. Parameter: Word. Returns: True or False.
def is_palindrome(word):
    # Convert the word to lowercase (for case insensitivity)
    word = word.lower()
    # Compare the reversed version of the word.
    return word == word[::-1]

### PROGRAM LOGIC ###

# Ask user for a word.
user_input = input("Enter a word: ")

# Call the function
result = is_palindrome(user_input)

# Print a user-friendly message for the result.
if result:
    print(f"{user_input} is a palindrome!")
else:
    print(f"{user_input} is not a palindrome.")
```
