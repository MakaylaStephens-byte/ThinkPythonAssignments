```python
# NAME: rentalcar.py

### IMPORT STATEMENTS ###
# (none)

### FUNCTION DEFINITIONS ###

# 1. Create a function: validate_class. Parameters: class code. Return: upper_code or none
def validate_class(class_code):
  # Checks for correct car classification code.
  upper_code = class_code.strip().upper()
  if upper_code == 'B' or upper_code == 'S' or upper_code == 'L':
    return upper_code
  else:
    return None
# 2. Create a function: get_class_name. Parameters: full code name. Return: full name or none
def get_class_name(full_code):
  # Gives the full names of each car classification.
  if full_code == 'B':
    return 'Budget'
  elif full_code == 'S':
    return 'Standard'
  elif full_code == 'L':
    return 'Luxury'
  else:
    return None

# 3. Create a function: calculate_mileage_charge. Parameters: miles driven  Return: charged_miles, charge_amount
def calculate_mileage_charge(miles_driven):
  # Calculates how many miles are subject to the mileage fee.

    # if miles driven is over 100 miles, the fee is charged
    if miles_driven > 100:
      charged_miles = miles_driven - 100
      charge_amount = charged_miles * 0.35

    # otherwise, it's free
    else:
      charged_miles = 0
      charge_amount = 0

    # return milage charge, charged amount
    return charged_miles, charge_amount

# 4. Create a function: calculate_total_bill. Parameters: class code, days, miles.  Return: total_bill, class_name, charged_miles
def calculate_total_bill(class_code, days, miles):
  # Calculate the total rental bill.
  validated = validate_class(class_code)
  class_name = get_class_name(validated)

  # The base rates
  rates = { 'B': 40, 'S': 60, 'L': 80}
  base_rate = rates[validated]

  # Calculate daily cost
  daily_cost = base_rate * days

  # Calculate milage cost
  charge_miles, milage_cost = calculate_mileage_charge(miles)

  # Return total_bill, class_name, charged_miles
  total_bill = daily_cost + milage_cost
  return total_bill, class_name, charge_miles

### PROGRAM LOGIC ###

# Input car classification
while True:
  class_input = input("Enter car classification (B, S, or L): ")
  class_code = validate_class(class_input)
  if class_code:
    break
  else:
    print("Invalid car classification. Please try again.")

# Input rental days
while True:
  days = int(input("Enter number of rental days: "))
  if days > 0:
    break
  else:
    print("Invalid number of days. Please try again.")

# Input odometer readings and calculate distance
while True:
  start_odometer = float(input("Enter starting odometer reading: "))
  end_odometer = float(input("Enter ending odometer reading: "))
  if end_odometer >= start_odometer:
    miles_driven = end_odometer - start_odometer
    break
  else:
    print("Invalid odometer readings. Please try again.")

# Call the function
total_bill, class_name, charged_miles = calculate_total_bill(class_code, days, miles_driven)

# Display final results
print("\n--- Rental Summary ---")
print(f"Car Classification: {class_name}")
print(f"Miles Driven:       {miles_driven:.1f} miles")
print(f"Charged Miles:      {charged_miles:.1f} miles")
print(f"Total Bill Amount:  ${total_bill:.2f}")
```
