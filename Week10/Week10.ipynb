```python
# NAME: roster.py

# IMPORT STATEMENTS

# (none)

# FUNCTION DEFINITIONS

###### First function: add_players. Parameter: player_list (should accept a list) ######
def add_players(player_list):

  # As long as the user is able to enter player names one by one when prompted.
  # The function also stops when user enters "quit".
  while True:
    name = input("Enter a player's name (or type 'quit' to finish): ").strip()
    if name == "quit":
      break
    player_list.append(name)



###### Second function: is_on_roster. Parameter: player_list, search_name. ######
def is_on_roster(player_list, search_name):
  # Checks to see if player's name is in the list.
  # Return true, otherwise false.
  if search_name in player_list:
    return True
  else:
    return False


# PROGRAM LOGIC

# Create an empty list called roster.
roster = []

# Call add_players to fill the list.
add_players(roster)

# Ask the user for one specific name to search for.
search_name = input("Enter a name to search for: ").strip()

# Call is_on_roster and use the returned value to print a message.
if is_on_roster(roster, search_name):
    print("Player found!")
else:
    print("Player not found.")
```
