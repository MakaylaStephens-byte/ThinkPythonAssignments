```python
# NAME: rectangle.py

# IMPORT STATEMENTS
# (NONE)

# FUNCTION DEFINITIONS

# 1.) Function: calculate_rectangle Parameters: length and width
def calculate_rectangle(length, width):

# Formula to calculate area
  area = length * width

# Formula to calculate perimeter
  perimeter = 2 * (length + width)

# Returns: A tuple
  return (area, perimeter)


# PROGRAM LOGIC

# Ask user to for length and width input (floats)
length_input = float(input("Enter the length: "))
width_input = float(input("Enter the width: "))

# Call the function while upacking the returned tuple
my_area, my_perimeter = calculate_rectangle(length_input, width_input)

# Print results
print("Results: ")
print("The Area is: ", my_area)
print("The Perimeter is: ", my_perimeter)
```
